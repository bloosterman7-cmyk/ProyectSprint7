[Calcula los ingresos mensuales por usuario (resta el límite del paquete gratuito del número total de llamadas, mensajes de texto y datos; multiplica el resultado por el valor del plan de llamadas; añade la tarifa mensual en función del plan de llamadas).
```python
# Calcula el ingreso mensual para cada usuario

def calcular_ingresos(row):
    plan = row['Plan']
    mins = row['monthly_minutes']
    sms = row['monthly_messages_sent']
    gb = np.ceil(row['monthly_Gb_spent']) #Aquí si se divide entre 1024 y se hacen las operaciones posteriores el valor es más alto. Es algo sobre el tipo de dato pero me quedé con esa duda
    if plan =='surf':
        costo = 20
        if mins>=500:
            costo += (mins-500)*0.03
        if sms>=50:
            costo += (sms-50)*0.03
        if gb>=15:
            costo += (gb-15)*10
    elif plan == 'ultimate':
        costo = 70
        if mins>=3000:
            costo += (mins-3000)*0.01
        if sms>=1000:
            costo += (sms-1000)*0.01
        if gb>=300:
            costo += (gb-300)*7
    else:
        costo = 0
    return costo

df['income'] = df.apply(calcular_ingresos, axis=1)
df.head(15)
```

[Prueba la hipótesis de que son diferentes los ingresos promedio procedentes de los usuarios de los planes de llamada Ultimate y Surf.]

[Elabora las hipótesis nula y alternativa, escoge la prueba estadística, determina el valor alfa.]

Hipótesis Nula:
Los ingresos promedio de los usuarios de los planes Ultimate y Surf son iguales.

Hipótesis Alternativa: 
Los ingresos promedio de los usuarios de los planes Ultimate y Surf son diferentes. 

2. Parámetros y Prueba
Nivel de significancia: 0.05 (5%). Este es el estándar en análisis de datos; si el valor p es menor a 0.05, rechazamos 


<div class="alert alert-block alert-danger">
<b>Reviewer's comment, Iteration 1</b> <a class="tocSkip"></a>

Tu redacción de la hipótesis está incompleta: indica claramente un planteamiento bilateral (ya que pruebas “diferente”), especifica la regla de decisión usando alpha (p. ej., rechazar H0 si p-value < alpha) y justifica brevemente por qué el Welch’s two-sample t-test es apropiado para estas muestras de ingresos.

</div>


```python
# Prueba las hipótesis
from scipy import stats

# Filtramos los ingresos por plan
ingresos_surf = df[df['Plan'] == 'surf']['income']
ingresos_ultimate = df[df['Plan'] == 'ultimate']['income']

# Realizamos la prueba T de muestras independientes
# equal_var=False porque no asumimos que las varianzas sean iguales
results = stats.ttest_ind(ingresos_surf, ingresos_ultimate, equal_var=False)

alpha = 0.05

print(f"Valor p: {results.pvalue}")

if results.pvalue < alpha:
    print("Rechazamos la hipótesis nula: Hay una diferencia significativa entre los ingresos.")
else:
    print("No podemos rechazar la hipótesis nula: No hay evidencia de una diferencia significativa.")
```

[Prueba la hipótesis de que el ingreso promedio de los usuarios del área NY-NJ es diferente al de los usuarios de otras regiones.]

[Elabora las hipótesis nula y alternativa, escoge la prueba estadística, determina el valor alfa.]

Definición de Hipótesis
Hipótesis Nula: El ingreso promedio de los usuarios en el área de NY-NJ-PA es igual al ingreso promedio de los usuarios de otras regiones.

Hipótesis Alternativa: El ingreso promedio de los usuarios en el área de NY-NJ-PA es diferente al de otras regiones.

Prueba Estadística: Prueba T de Student para dos muestras independientes. Dado que el tamaño de los grupos y sus varianzas suelen ser desiguales, utilizaremos la Prueba de Welch (equal_var=False).
Valor Alfa: (Nivel de confianza del 95%).
     



```python
# Prueba las hipótesis

from scipy import stats

# 1. Unir la columna 'area' de df5 a df usando 'user_id' como llave
df = df.merge(df5[['user_id', 'area']], on='user_id', how='left')

# 2. Definir los dos grupos de ingresos
area_target = 'New York-Newark-Jersey City, NY-NJ-PA MSA'
ingresos_ny_nj = df[df['area'] == area_target]['income']
ingresos_otros = df[df['area'] != area_target]['income']

# 3. Realizar la prueba T de Welch
results = stats.ttest_ind(ingresos_ny_nj, ingresos_otros, equal_var=False)

print(f"Valor p: {results.pvalue}")

# 4. Decisión
if results.pvalue < 0.05:
    print("Rechazamos la hipótesis nula: Hay una diferencia significativa en los ingresos.")
else:
    print("No podemos rechazar la hipótesis nula: No hay evidencia de que los ingresos sean diferentes.")
```
